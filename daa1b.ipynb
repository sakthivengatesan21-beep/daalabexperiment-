```python
import time
import random
import sys
def binary_search(arr, x):
    low = 0
    high = len(arr) - 1
    probes = 0

    while low <= high:
        probes += 1
        mid = (low + high) // 2

        if arr[mid] == x:
            return mid, probes
        elif arr[mid] < x:
            low = mid + 1
        else:
            high = mid - 1

    return -1, probes


# Driver Code
if __name__ == "__main__":
    # Sorted roll numbers (range 1–10000)
    roll_numbers = list(range(1, 10001))

    # Multiple targets to search
    targets = [1, 500, 1234, 5678, 9876, 10000, 11111]  # last one is not in array

    # Print table header
    print("{:<12} {:<12} {:<12} {:<12}".format("Roll Number", "Index Found", "Probes", "Status"))
    print("-" * 50)

    # Run Binary Search for each target
    for target in targets:
        pos, probes = binary_search(roll_numbers, target)
        if pos != -1:
            print("{:<12} {:<12} {:<12} {:<12}".format(target, pos, probes, "Found"))
        else:
            print("{:<12} {:<12} {:<12} {:<12}".format(target, "-", probes, "Not Found"))
```
